###  Recipe Scraper Function — `scrape_recipe_details(url)`

This function is responsible for **scraping complete recipe details** from a given webpage URL.  
It uses **BeautifulSoup** to parse the HTML and extracts key recipe properties such as title, ingredients, cooking steps, and nutrition data.
It’s part of the testing phase to validate Nutrition API integration before full pipeline implementation.


```python
def scrape_recipe_details(url):
    print(f" Fetching HTML from: {url}")
    headers = {"User-Agent": "Mozilla/5.0 (compatible; RecipeScraper/2.0)"}
    html = requests.get(url, headers=headers, timeout=20).text
    soup = BeautifulSoup(html, "lxml")

    # initialize recipe data structure 
    recipe_data = {
        "source_url": url,
        "title": None,
        "description": None,
        "cuisine_type": None,
        "prep_steps": [],
        "prep_time": None,
        "cook_steps": [],
        "cook_time": None,
        "meal_type": None,
        "course_type": None,
        "youtube_link": None,
        "serving_count": None,
        "quantity_ingredients": [],
        "nutrition": None  # will fill below
    }

    #  extract JSON-LD data 
    for script in soup.find_all("script", type="application/ld+json"):
        try:
            data = json.loads(script.string.strip())

            # handle arrays and graphs
            if isinstance(data, list):
                data = next((d for d in data if isinstance(d, dict) and d.get("@type") == "Recipe"), None)
            elif isinstance(data, dict) and "@graph" in data:
                data = next((g for g in data["@graph"] if g.get("@type") == "Recipe"), None)

            if not data or data.get("@type") != "Recipe":
                continue

            # extract basic info
            recipe_data["title"] = data.get("name")
            recipe_data["description"] = data.get("description")
            recipe_data["cuisine_type"] = data.get("recipeCuisine")
            recipe_data["prep_time"] = data.get("prepTime")
            recipe_data["cook_time"] = data.get("cookTime")
            recipe_data["serving_count"] = data.get("recipeYield")
            recipe_data["course_type"] = data.get("recipeCategory")
            recipe_data["quantity_ingredients"] = data.get("recipeIngredient", [])

            # extract instructions
            instructions = data.get("recipeInstructions", [])
            steps = []

            def extract_steps(obj):
                if isinstance(obj, list):
                    for o in obj:
                        extract_steps(o)
                elif isinstance(obj, dict):
                    if "text" in obj:
                        steps.append(obj["text"])
                    if "itemListElement" in obj:
                        extract_steps(obj["itemListElement"])
                elif isinstance(obj, str):
                    steps.append(obj)

            extract_steps(instructions)
            mid = len(steps) // 2
            recipe_data["prep_steps"] = steps[:mid]
            recipe_data["cook_steps"] = steps[mid:]

            break  # stop after first Recipe found

        except Exception:
            continue

    # YouTube 
    iframe = soup.find("iframe")
    if iframe and "youtube" in iframe.get("src", ""):
        recipe_data["youtube_link"] = iframe["src"]

    # Nutrition Extraction
    ingredients = recipe_data.get("quantity_ingredients", [])
    if ingredients:
        print(" Fetching nutrition data from Nutritionix API...")
        nutrition = get_nutrition_from_nutritionix(ingredients)
        recipe_data["nutrition"] = nutrition or {
            "calories_kcal": None,
            "carbohydrates_g": None,
            "protein_g": None,
            "fat_g": None,
            "fiber_g": None,
            "sodium_mg": None
        }
    else:
        print(" No ingredients found for nutrition extraction.")
        recipe_data["nutrition"] = {
            "calories_kcal": None,
            "carbohydrates_g": None,
            "protein_g": None,
            "fat_g": None,
            "fiber_g": None,
            "sodium_mg": None
        }

    return recipe_data
```

```python
# Call the get_recipe_urls_google with desired parameters to obtain the web URLs
urls = get_recipe_urls_google("American", "Chocolate cake", max_links=3)

# Using json.dumps() function all the details are extracted in the defined properties
all_recipes = []
for url in urls:
    recipe = scrape_recipe_details(url)
    print(json.dumps(recipe, indent=2, ensure_ascii=False))
    all_recipes.append(recipe)
```

## Ingredient and Grocery Mapping 

The ingredient and grocery mapping system helps in **standardizing recipe data** by converting all ingredient quantities into grams, and then mapping each ingredient to its **nearest grocery equivalent**.  


### Unit Conversion Logic

The system defines **base unit conversions** for three main types of quantities:

- **Mass units**:  
  `1 kg = 1000 g`, `1 lb = 453.592 g`, `1 oz = 28.3495 g`
- **Volume units**:  
  `1 cup = 240 ml`, `1 tbsp = 15 ml`, `1 tsp = 5 ml`, `1 l = 1000 ml`
- **Count units**:  
  Treated as individual items (`1 piece`, `1 clove`, `1 bunch`, etc.)

Each unit is mapped to its **base SI equivalent (grams or milliliters)** for accurate calculations.


### Ingredient Database

A built-in **ingredient reference database** stores predefined details for common items such as onion, garlic, flour, butter, sugar, and tomato.  
Each entry includes:

- **Density (g/ml):** Used to convert volume to mass.  
- **Average piece weight (g):** Used for count-based ingredients.  
- **Yield fraction:** Accounts for usable portion (e.g., after peeling or trimming).  
- **Grocery name and unit:** How the item is sold in stores.  
- **MOQ (Minimum Order Quantity):** Minimum purchasable unit (e.g., 1 kg or 1 pack).  
- **Shelf life:** Approximate freshness duration in days.  

If an ingredient isn’t found in the database, a **default category-based profile** is applied (e.g., vegetable, dry, or liquid).


### Ingredient Normalization

Before mapping, ingredient names are **cleaned and normalized** by:
- Removing descriptors like *chopped, diced, fresh, peeled, ground*, etc.  
- Converting text to lowercase and stripping special characters.  
- Using **fuzzy matching** to find close matches in the database (e.g., “onions” -> “onion”).  
- If no match is found, it assigns a **default category** such as “vegetable”, “dry”, or “liquid”.


### Conversion to Grams

Each recipe ingredient is converted to grams using its **unit type**:

- **Mass units:**  
  Direct conversion (e.g., 2 kg -> 2000 g)
- **Volume units:**  
  Converted using density (e.g., 1 cup of oil -> 240 ml × 0.91 g/ml = 218 g)
- **Count units:**  
  Uses the average piece weight (e.g., 2 onions -> 2 × 150 g = 300 g)

This ensures all ingredients are represented in a **uniform mass-based system (grams)**.


### Grocery Mapping

After standardization, each ingredient is **mapped to its corresponding grocery item**, determining how much of the grocery unit is required.  

For example:
- Recipe calls for **300 g onion**  
- Grocery pack: **1 kg onions (usable yield 0.92)**  
- Usable mass = 1000 * 0.92 = 920 g  
- **Conversion ratio:** 920 / 300 ~= 3.07  
- **Grocery quantity needed:** 300 / 920 = 0.33 kg  

Hence, **0.33 kg of onion** (or roughly 1/3 of a 1 kg pack) is required.

### Conversion Ratio

The **conversion ratio** represents how much usable grocery quantity compares to the recipe requirement:

$$
\text{Conversion Ratio} = \frac{\text{Cooking Quantity (g)}}{\text{Grocery Quantity Purchased (g)}}
$$

If:
- Conversion ratio > 1 -> One grocery unit provides more than needed  
- Conversion ratio < 1 -> Multiple grocery units are needed  

It helps in **inventory planning** and **purchase estimation** for batch cooking or grocery ordering.

### Output Data Structure

Each mapped ingredient includes:
- Ingredient name (raw form)  
- Recipe quantity and unit  
- Grocery name and unit  
- Grocery quantity needed  
- Conversion ratio  
- MOQ and shelf life  
- Notes (fuzzy match or inferred category)

```python
# Base conversions 
UNIT_CONVERSIONS = {
    "mass": {"g": 1, "kg": 1000, "oz": 28.3495, "lb": 453.592},
    "volume": {"tsp": 5, "tbsp": 15, "cup": 240, "ml": 1, "l": 1000},
    "count": {"piece": 1, "item": 1, "clove": 1, "bunch": 1}
}

# Known ingredient reference
INGREDIENT_DATABASE = {
    "onion": {"density_g_per_ml": 0.52, "avg_piece_g": 150, "yield_fraction": 0.92, "grocery_name": "Yellow Onions", "grocery_unit": "kg", "grocery_mass_g": 1000, "MOQ": 1, "MOQ_unit": "kg", "shelf_life_days": 2},
    "garlic": {"density_g_per_ml": 0.6, "avg_piece_g": 4, "yield_fraction": 0.95, "grocery_name": "Garlic Bulb", "grocery_unit": "kg", "grocery_mass_g": 1000, "MOQ": 0.5, "MOQ_unit": "kg", "shelf_life_days": 7},
    "flour": {"density_g_per_ml": 0.53, "yield_fraction": 1.0, "grocery_name": "All-purpose Flour", "grocery_unit": "kg", "grocery_mass_g": 1000, "MOQ": 1, "MOQ_unit": "kg", "shelf_life_days": 90},
    "butter": {"density_g_per_ml": 0.911, "yield_fraction": 1.0, "grocery_name": "Butter", "grocery_unit": "pack", "grocery_mass_g": 500, "MOQ": 1, "MOQ_unit": "pack", "shelf_life_days": 30},
    "sugar": {"density_g_per_ml": 0.85, "yield_fraction": 1.0, "grocery_name": "Sugar", "grocery_unit": "kg", "grocery_mass_g": 1000, "MOQ": 1, "MOQ_unit": "kg", "shelf_life_days": 365},
    "tomato": {"density_g_per_ml": 0.95, "avg_piece_g": 120, "yield_fraction": 0.9, "grocery_name": "Tomatoes", "grocery_unit": "kg", "grocery_mass_g": 1000, "MOQ": 1, "MOQ_unit": "kg", "shelf_life_days": 3},
}

# Default fallback profiles (category-based) 
DEFAULTS = {
    "vegetable": {"density_g_per_ml": 0.6, "avg_piece_g": 100, "yield_fraction": 0.9, "grocery_unit": "kg", "grocery_mass_g": 1000, "MOQ": 1, "MOQ_unit": "kg", "shelf_life_days": 3},
    "liquid": {"density_g_per_ml": 1.0, "yield_fraction": 1.0, "grocery_unit": "l", "grocery_mass_g": 1000, "MOQ": 1, "MOQ_unit": "l", "shelf_life_days": 7},
    "dry": {"density_g_per_ml": 0.5, "yield_fraction": 1.0, "grocery_unit": "kg", "grocery_mass_g": 1000, "MOQ": 1, "MOQ_unit": "kg", "shelf_life_days": 180},
}

# Utility functions
def find_unit_type(unit: str) -> str:
    for t, mapping in UNIT_CONVERSIONS.items():
        if unit in mapping:
            return t
    return "count"

def normalize_name(name: str) -> str:
    name = name.lower()
    name = re.sub(r"[\d\W_]+", " ", name)
    name = re.sub(r"\b(chopped|sliced|diced|minced|fresh|whole|ground|powdered|peeled)\b", "", name)
    return name.strip()

def guess_category(name: str) -> str:
    if any(w in name for w in ["milk", "oil", "water", "sauce", "vinegar"]):
        return "liquid"
    if any(w in name for w in ["flour", "sugar", "rice", "powder", "salt"]):
        return "dry"
    return "vegetable"

def get_ingredient_profile(raw_name: str) -> Dict[str, Any]:
    clean_name = normalize_name(raw_name)
    if clean_name in INGREDIENT_DATABASE:
        return INGREDIENT_DATABASE[clean_name]
    close = get_close_matches(clean_name, INGREDIENT_DATABASE.keys(), n=1, cutoff=0.7)
    if close:
        base = INGREDIENT_DATABASE[close[0]].copy()
        base["grocery_name"] = close[0].title()
        base["note"] = f"Fuzzy matched to '{close[0]}'"
        return base
    cat = guess_category(clean_name)
    base = DEFAULTS[cat].copy()
    base["grocery_name"] = f"Generic {cat.title()}"
    base["note"] = f"Inferred from category '{cat}'"
    return base

def recipe_to_grams(ingredient_data: Dict[str, Any], qty: float, unit: str) -> float:
    density = ingredient_data.get("density_g_per_ml", 1.0)
    piece_g = ingredient_data.get("avg_piece_g", 100)
    utype = find_unit_type(unit)
    if utype == "mass":
        return qty * UNIT_CONVERSIONS["mass"][unit]
    elif utype == "volume":
        return qty * UNIT_CONVERSIONS["volume"][unit] * density
    else:
        return qty * piece_g

def map_ingredient_to_grocery(raw_name: str, qty: float, unit: str) -> Dict[str, Any]:
    data = get_ingredient_profile(raw_name)
    g_mass_g = data["grocery_mass_g"]
    yield_y = data["yield_fraction"]

    mass_needed_g = recipe_to_grams(data, qty, unit)
    available_g = g_mass_g * yield_y
    cr = available_g / mass_needed_g if mass_needed_g else None
    grocery_units_needed = mass_needed_g / available_g if available_g else None

    return {
        "ingredient_name": raw_name,
        "recipe_qty": qty,
        "recipe_unit": unit,
        "grocery_name": data["grocery_name"],
        "grocery_unit": data["grocery_unit"],
        "conversion_ratio": round(cr, 2) if cr else None,
        "grocery_qty_needed": round(grocery_units_needed, 3) if grocery_units_needed else None,
        "MOQ": f"{data['MOQ']} {data['MOQ_unit']}",
        "shelf_life_days": data["shelf_life_days"],
        "note": data.get("note", "")
    }
```

## Enhanced Recipe Scraper

The **Enhanced Recipe Scraper** is designed to automatically extract, clean, and enrich recipe data from web pages.  
It retrieves essential information such as title, ingredients, instructions, nutrition, and YouTube video link, and further performs ingredient-to-grocery mapping for standardized data storage.

### Workflow Overview

- It parses the webpage with **BeautifulSoup** for structured data extraction.
- The parsed HTML is analyzed to find `<script>` tags containing **JSON-LD** structured recipe data (`@type: Recipe`).

  - **Title, Description, Cuisine, and Course Type**

  - Fat  
  - Fiber  
  - Sodium  

If no ingredients are available, default null values are used.

###  Final Structured Data

The final output is a well-structured dictionary containing:

- Ingredient-to-grocery mappings

- Embedded YouTube video links (if available)

###  Summary

This pipeline makes recipe data **searchable, analyzable, and scalable** for real-world applications.


```python
# Enhanced recipe scraper
def scrape_recipe_details(url):
    print(f" Fetching HTML from: {url}")
    headers = {"User-Agent": "Mozilla/5.0 (compatible; RecipeScraper/2.0)"}
    html = requests.get(url, headers=headers, timeout=20).text
    soup = BeautifulSoup(html, "lxml")

    recipe_data = {
        "source_url": url,
        "title": None,
        "description": None,
        "cuisine_type": None,
        "prep_steps": [],
        "prep_time": None,
        "cook_steps": [],
        "cook_time": None,
        "meal_type": None,
        "course_type": None,
        "youtube_link": None,
        "serving_count": None,
        "nutrition": None,
        "quantity_ingredients": [],
        "mapped_ingredients": []
    }

    for script in soup.find_all("script", type="application/ld+json"):
        try:
            data = json.loads(script.string.strip())
            if isinstance(data, list):
                data = next((d for d in data if isinstance(d, dict) and d.get("@type") == "Recipe"), None)
            elif isinstance(data, dict) and "@graph" in data:
                data = next((g for g in data["@graph"] if g.get("@type") == "Recipe"), None)
            if not data or data.get("@type") != "Recipe":
                continue

            recipe_data["title"] = data.get("name")
            recipe_data["description"] = data.get("description")
            recipe_data["cuisine_type"] = data.get("recipeCuisine")
            recipe_data["prep_time"] = data.get("prepTime")
            recipe_data["cook_time"] = data.get("cookTime")
            recipe_data["serving_count"] = data.get("recipeYield")
            recipe_data["course_type"] = data.get("recipeCategory")
            recipe_data["quantity_ingredients"] = data.get("recipeIngredient", [])

            instructions = data.get("recipeInstructions", [])
            steps = []

            def extract_steps(obj):
                if isinstance(obj, list):
                    for o in obj:
                        extract_steps(o)
                elif isinstance(obj, dict):
                    if "text" in obj:
                        steps.append(obj["text"])
                    if "itemListElement" in obj:
                        extract_steps(obj["itemListElement"])
                elif isinstance(obj, str):
                    steps.append(obj)

            extract_steps(instructions)
            mid = len(steps) // 2
            recipe_data["prep_steps"] = steps[:mid]
            recipe_data["cook_steps"] = steps[mid:]
            break
        except Exception:
            continue

    iframe = soup.find("iframe")
    if iframe and "youtube" in iframe.get("src", ""):
        recipe_data["youtube_link"] = iframe["src"]

    iframe = soup.find("iframe")
    if iframe and "youtube" in iframe.get("src", ""):
        recipe_data["youtube_link"] = iframe["src"]

    # Nutrition Extraction
    ingredients = recipe_data.get("quantity_ingredients", [])
    if ingredients:
        print(" Fetching nutrition data from Nutritionix API...")
        nutrition = get_nutrition_from_nutritionix(ingredients)
        recipe_data["nutrition"] = nutrition or {
            "calories_kcal": None,
            "carbohydrates_g": None,
            "protein_g": None,
            "fat_g": None,
            "fiber_g": None,
            "sodium_mg": None
        }
    else:
        print(" No ingredients found for nutrition extraction.")
        recipe_data["nutrition"] = {
            "calories_kcal": None,
            "carbohydrates_g": None,
            "protein_g": None,
            "fat_g": None,
            "fiber_g": None,
            "sodium_mg": None
        }

    # Map ingredients to groceries
    ingredients = recipe_data.get("quantity_ingredients", [])
    mapped = []
    for ing in ingredients:
        # Trying to parse quantity and unit roughly (we can enhance this with regex)
        match = re.match(r"([\d\.\/]+)?\s*(\w+)?\s*(.*)", ing)
        qty = eval(match.group(1)) if match and match.group(1) else 1
        unit = match.group(2).lower() if match and match.group(2) else "piece"
        name = match.group(3)
        mapped.append(map_ingredient_to_grocery(name, qty, unit))
    recipe_data["mapped_ingredients"] = mapped

    return recipe_data

```

```python
import re
import json
import time

def normalize_recipe_generic(recipe: dict) -> dict:
    """
    Generalized normalization for any scraped recipe dictionary.
    Ensures consistent keys, default values, and structure.
    """

    def safe_get(d, k, default=None):
        return d[k] if k in d and d[k] not in [None, ""] else default

    # Meta data
    meta = {
        "title": safe_get(recipe, "title", "Untitled Recipe"),
        "source_url": safe_get(recipe, "source_url"),
        "description": safe_get(recipe, "description", ""),
        "cuisine_type": safe_get(recipe, "cuisine_type", "Unknown"),
        "course_type": safe_get(recipe, "course_type", "Unknown"),
        "meal_type": safe_get(recipe, "meal_type", "Unknown"),
        "serving_count": safe_get(recipe, "serving_count", 1),
        "prep_time": safe_get(recipe, "prep_time", "N/A"),
        "cook_time": safe_get(recipe, "cook_time", "N/A"),
        "youtube_link": safe_get(recipe, "youtube_link"),
        "timestamp": time.strftime("%Y-%m-%d %H:%M:%S"),
    }

    # Ingredient Normalization
    ingredients = []
    for i in recipe.get("mapped_ingredients") or recipe.get("ingredient_mappings") or []:
        ingredients.append({
            "ingredient_name": safe_get(i, "ingredient_name", "Unknown Ingredient"),
            "recipe_quantity": {
                "value": safe_get(i, "recipe_qty", 1.0),
                "unit": safe_get(i, "recipe_unit", "unit"),
            },
            "grocery_mapping": {
                "grocery_name": safe_get(i, "grocery_name", "Generic Grocery"),
                "grocery_unit": safe_get(i, "grocery_unit", "unit"),
                "conversion_ratio": safe_get(i, "conversion_ratio", 1.0),
                "grocery_qty_needed": safe_get(i, "grocery_qty_needed", 1.0),
                "MOQ": safe_get(i, "MOQ", "1 unit"),
                "shelf_life_days": safe_get(i, "shelf_life_days", 7),
            },
            "note": safe_get(i, "note", "")
        })

    # Steps Section
    steps = {
        "prep": [s for s in recipe.get("prep_steps", []) if isinstance(s, str) and s.strip()],
        "cook": [s for s in recipe.get("cook_steps", []) if isinstance(s, str) and s.strip()],
    }

    # Nutrition Section
    nutrition_raw = safe_get(recipe, "nutrition", {})
    nutrition = {
        "calories_kcal": safe_get(nutrition_raw, "calories_kcal", None),
        "carbohydrates_g": safe_get(nutrition_raw, "carbohydrates_g", None),
        "protein_g": safe_get(nutrition_raw, "protein_g", None),
        "fat_g": safe_get(nutrition_raw, "fat_g", None),
        "fiber_g": safe_get(nutrition_raw, "fiber_g", None),
        "sodium_mg": safe_get(nutrition_raw, "sodium_mg", None),
        "note": "Values are estimated via Nutritionix or similar API"
    }

    # Building a Final Normalized Structure
    normalized = {
        "meta": meta,
        "ingredients": ingredients,
        "steps": steps,
        "nutrition": nutrition,
    }

    return normalized
```

```python
# Extracting and Saving the normalized data into JSON files

def save_normalized_recipe(recipe, output_dir="structured_recipes"):
    """Normalize and save a single recipe JSON file."""
    os.makedirs(output_dir, exist_ok=True)
    normalized = normalize_recipe_generic(recipe)
    filename = re.sub(r"[^\w\s-]", "", normalized["meta"]["title"]).strip().replace(" ", "_") + ".json"
    path = os.path.join(output_dir, filename)
    with open(path, "w", encoding="utf-8") as f:
        json.dump(normalized, f, indent=2, ensure_ascii=False)
    print(f"✅ Normalized recipe saved to: {os.path.abspath(path)}")
    return normalized
```

```python
import os
import json
from pathlib import Path
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

# My Neon database connection
DATABASE_URL = 'postgresql://neondb_owner:npg_Leu95kAsaSHc@ep-silent-cake-a1yjujhc-pooler.ap-southeast-1.aws.neon.tech/neondb?sslmode=require&channel_binding=require'

# My Folder containing your JSON recipe files
JSON_FOLDER = Path(r"C:\Users\Heman\OneDrive\Desktop\Normalized Recipes")


TABLE_NAME = "recipes_normalized"

# Creating table 'recipe_normalized'
def create_table_if_not_exists(engine):
    create_sql = f"""
    CREATE TABLE IF NOT EXISTS {TABLE_NAME} (
        id SERIAL PRIMARY KEY,
        filename TEXT,
        title TEXT,
        description TEXT,
        cuisine_type TEXT,
        course_type TEXT,
        meal_type TEXT,
        youtube_video TEXT,
        servings_set TEXT,
        prep_time TEXT,
        cook_time TEXT,
        prep_steps JSONB,
        cook_steps JSONB,
        fat_g FLOAT,
        carbs_g FLOAT,
        fiber_g FLOAT,
        protein_g FLOAT,
        sodium_mg FLOAT,
        calories_kcal FLOAT,
        ingredients JSONB,
        grocery_mapping JSONB,
        loaded_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
    );
    """
    with engine.begin() as conn:
        conn.execute(text(create_sql))
    print(f"Table '{TABLE_NAME}' is ready.")

def extract_value(data, *keys):
    #Safely get nested values
    for key in keys:
        if isinstance(data, dict):
            data = data.get(key)
        else:
            return None
    return data

def insert_json_file(engine, filepath):
    with open(filepath, "r", encoding="utf-8") as f:
        data = json.load(f)
    filename = os.path.basename(filepath)

    # Extract fields safely
    title = extract_value(data, "title")
    description = extract_value(data, "meta", "description") or extract_value(data, "description")
    cuisine_type = extract_value(data, "meta", "cuisine_type", 0) or extract_value(data, "cuisine_type")
    course_type = extract_value(data, "meta", "course_type", 0) or extract_value(data, "course_type")
    meal_type = extract_value(data, "meal_type")
    youtube_video = extract_value(data, "youtube_video")
    servings_set = extract_value(data, "servings_set")
    prep_time = extract_value(data, "prep_time")
    cook_time = extract_value(data, "cook_time")
    prep_steps = extract_value(data, "prep_steps") or extract_value(data, "instructions", "prep")
    cook_steps = extract_value(data, "cook_steps") or extract_value(data, "instructions", "cook")

    nutrition = extract_value(data, "nutrition") or {}
    fat_g = nutrition.get("fat_g")
    carbs_g = nutrition.get("carbs_g")
    fiber_g = nutrition.get("fiber_g")
    protein_g = nutrition.get("protein_g")
    sodium_mg = nutrition.get("sodium_mg")
    calories_kcal = nutrition.get("calories_kcal")

    ingredients = extract_value(data, "ingredients") or []
    grocery_mapping = extract_value(data, "grocery_mapping") or {}

    insert_sql = text(f"""
        INSERT INTO {TABLE_NAME} (
            filename, title, description, cuisine_type, course_type, meal_type, youtube_video,
            servings_set, prep_time, cook_time, prep_steps, cook_steps,
            fat_g, carbs_g, fiber_g, protein_g, sodium_mg, calories_kcal,
            ingredients, grocery_mapping
        )
        VALUES (
            :filename, :title, :description, :cuisine_type, :course_type, :meal_type, :youtube_video,
            :servings_set, :prep_time, :cook_time, :prep_steps, :cook_steps,
            :fat_g, :carbs_g, :fiber_g, :protein_g, :sodium_mg, :calories_kcal,
            :ingredients, :grocery_mapping
        )
    """)

    with engine.begin() as conn:
        conn.execute(insert_sql, {
            "filename": filename,
            "title": title,
            "description": description,
            "cuisine_type": cuisine_type,
            "course_type": course_type,
            "meal_type": meal_type,
            "youtube_video": youtube_video,
            "servings_set": servings_set,
            "prep_time": prep_time,
            "cook_time": cook_time,
            "prep_steps": json.dumps(prep_steps),
            "cook_steps": json.dumps(cook_steps),
            "fat_g": fat_g,
            "carbs_g": carbs_g,
            "fiber_g": fiber_g,
            "protein_g": protein_g,
            "sodium_mg": sodium_mg,
            "calories_kcal": calories_kcal,
            "ingredients": json.dumps(ingredients),
            "grocery_mapping": json.dumps(grocery_mapping)
        })
    print(f"Inserted: {filename}")

def load_all_jsons_to_postgres():
    engine = create_engine(DATABASE_URL, pool_pre_ping=True)
    create_table_if_not_exists(engine)
    files = list(Path(JSON_FOLDER).glob("*.json"))

    print(f"Found {len(files)} JSON files, uploading...")
    for file in files:
        try:
            insert_json_file(engine, file)
        except Exception as e:
            print(f"Error inserting {file.name}: {e}")
    print("All JSON files uploaded successfully.")
    engine.dispose()

# Run it
load_all_jsons_to_postgres()
```
